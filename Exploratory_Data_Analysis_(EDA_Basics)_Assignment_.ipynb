## Interpretation:

* Average price shows the overall market value.
* Median price represents the middle house price.
* Standard deviation shows price variation in the market.

High standard deviation indicates large variation between cheap and luxury houses.

# **Q2. Which number of bedrooms is most common?**

```python id="t7p0zc"
common_bedrooms = df['bedrooms'].mode()[0]

print("Most Common Bedrooms:", common_bedrooms)
```
## **Interpretation:**

The most common bedroom count represents the most preferred house size among buyers.

---

# **Q3. Check if the price data is skewed.**

```python id="m9q5lr"
price_skewness = df['price'].skew()

print("Price Skewness:", price_skewness)
```
## **Interpretation:**

* Positive skewness means there are some extremely expensive houses.
* Real estate datasets usually show right-skewed distribution because luxury houses increase overall prices.

---

# **Q4. What is the average price for each number of bedrooms?**

```python id="g8r2xw"
avg_price_bedrooms = df.groupby('bedrooms')['price'].mean()

print(avg_price_bedrooms)
```
## **Interpretation:**

Generally, houses with more bedrooms have higher average prices.

This helps understand buyer demand and pricing trends.

---

# **Q5. What is the relationship between living area and price?**

(Use Area without Basement)

```python id="s4z7pe"
correlation = df['living_in_m2'].corr(df['price'])

print("Correlation:", correlation)
```
## **Interpretation:**

* Positive correlation means larger living areas generally have higher prices.
* Strong correlation indicates area is an important factor affecting house price.

# **Q6. Identify anomalies where houses have high prices but low areas.**

```python id="v0n8rw"
anomalies = df[
    (df['price'] > df['price'].quantile(0.90)) &
    (df['living_in_m2'] < df['living_in_m2'].quantile(0.25))
]

print(anomalies[['price', 'living_in_m2']].head())
```
## **Interpretation:**

These houses may be expensive because of:

* prime location
* luxury interiors
* waterfront
* commercial value

Even with smaller area, some properties have very high prices.

---

# **Q7. Compare average price based on number of floors and houses with or without waterfront.**

```python id="u4x6jb"
floor_waterfront = df.groupby(
    ['floors', 'waterfront']
)['price'].mean()

print(floor_waterfront)
```
## **Interpretation:**

* Waterfront houses generally have higher prices.
* Multi-floor houses are usually more expensive.
* Luxury features significantly impact property value.

# **Q8. Identify the minimum and maximum house price. What does this indicate?**
```python id="k8m1tr"
minimum_price = df['price'].min()

maximum_price = df['price'].max()

print("Minimum Price:", minimum_price)

print("Maximum Price:", maximum_price)
```
## **Interpretation:**

* Minimum price represents affordable properties.
* Maximum price represents luxury or premium properties.

Large price differences indicate high diversity in the housing market.

---

# **Q9. Which location (zipcode/area) has the highest average price?**
```python id="e5p9yl"
location_price = df.groupby('zipcode')['price'].mean()

highest_location = location_price.idxmax()

highest_price = location_price.max()

print("Highest Average Price Zipcode:", highest_location)

print("Average Price:", highest_price)
```
## **Interpretation:**

This location is likely:

* premium area
* better infrastructure
* high demand area
* luxury housing zone
